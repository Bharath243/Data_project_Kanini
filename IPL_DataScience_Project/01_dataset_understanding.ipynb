Step 2: Unzip the raw Cricsheet IPL dataset.

```python
import zipfile, os

with zipfile.ZipFile('ipl_csv2.zip', 'r') as z:
    z.extractall('ipl_data')

files = os.listdir('ipl_data')
print(f"Total files: {len(files)}")
print(files[:10])
```

Step 7: Load the combined all_matches.csv file (every delivery, every match).

```python
all_matches = pd.read_csv('ipl_data/all_matches.csv')

print(all_matches.shape)                       # (rows, columns)
print(sorted(all_matches['season'].astype(str).unique()))  # confirm season coverage
print(all_matches['match_id'].nunique(), "unique matches")
```
